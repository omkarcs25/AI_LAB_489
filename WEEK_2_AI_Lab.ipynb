```python
import heapq

print("NAME: OMKAR SADASHIV KOKATNOOR")
print("USN: 1BM25CS489")
class PuzzleState:
    def __init__(self, board, g=0, parent=None, move=""):
        self.board = tuple(board)
        self.g = g
        self.parent = parent
        self.move = move
        self.h = self.calculate_manhattan()
        self.f = self.g + self.h

    def calculate_manhattan(self):
        goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)
        distance = 0
        for i in range(9):
            val = self.board[i]
            if val != 0:
                target_idx = goal.index(val)
                r1, c1 = divmod(i, 3)
                r2, c2 = divmod(target_idx, 3)
                distance += abs(r1 - r2) + abs(c1 - c2)
        return distance

    def get_neighbors(self):
        neighbors = []
        zero_idx = self.board.index(0)
        r, c = divmod(zero_idx, 3)

        moves = {
            "Up": (-1, 0),
            "Down": (1, 0),
            "Left": (0, -1),
            "Right": (0, 1)
        }

        for direction, (dr, dc) in moves.items():
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                neighbor_idx = nr * 3 + nc
                new_board = list(self.board)
                new_board[zero_idx], new_board[neighbor_idx] = new_board[neighbor_idx], new_board[zero_idx]
                neighbors.append(PuzzleState(new_board, self.g + 1, self, direction))
        return neighbors

    def __lt__(self, other):
        return self.f < other.f

def solve_puzzle(start_board, goal_board):
    start_state = PuzzleState(start_board)
    open_set = []
    heapq.heappush(open_set, start_state)
    closed_set = set()

    while open_set:
        current = heapq.heappop(open_set)

        if current.board == tuple(goal_board):
            path = []
            while current.parent:
                path.append((current.move, current.board))
                current = current.parent
            return path[::-1]

        closed_set.add(current.board)

        for neighbor in current.get_neighbors():
            if neighbor.board in closed_set:
                continue
            heapq.heappush(open_set, neighbor)

    return None

def print_board(board):
    for i in range(0, 9, 3):
        print(board[i:i+3])
    print()

initial_state = [1, 2, 3, 0, 4, 6, 7, 5, 8]
goal_state = [1, 2, 3, 4, 5, 6, 7, 8, 0]

print("Initial State:")
print_board(initial_state)

solution = solve_puzzle(initial_state, goal_state)
if solution:
    print(f"Solved in {len(solution)} steps!\n")
    for step, (move, board) in enumerate(solution, 1):
        print(f"Step {step}: Move 0 {move}")
        print_board(board)
else:
    print("No solution found.")
```

```python
print("NAME: OMKAR SADASHIV KOKATNOOR")
print("USN: 1BM25CS489")
def standard_dfs_solve(start_board, goal_board, max_depth=15):
    goal_tuple = tuple(goal_board)
    visited = set()

    def dfs(board, path):
        if board == goal_tuple:
            return path
        if len(path) >= max_depth:
            return None

        zero_idx = board.index(0)
        r, c = divmod(zero_idx, 3)

        # Standard DFS move order: Up, Down, Left, Right
        moves = {
            "Up": (-1, 0),
            "Down": (1, 0),
            "Left": (0, -1),
            "Right": (0, 1)
        }

        for direction, (dr, dc) in moves.items():
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                neighbor_idx = nr * 3 + nc
                new_board = list(board)
                new_board[zero_idx], new_board[neighbor_idx] = new_board[neighbor_idx], new_board[zero_idx]
                new_board_tuple = tuple(new_board)

                if new_board_tuple not in visited:
                    visited.add(new_board_tuple)
                    result = dfs(new_board_tuple, path + [(direction, new_board_tuple)])
                    if result is not None:
                        return result
                    visited.remove(new_board_tuple)
        return None

    visited.add(tuple(start_board))
    return dfs(tuple(start_board), [])

# Let's run pure DFS on the same states
initial_state = [1, 2, 3, 0, 4, 6, 7, 5, 8]
goal_state = [1, 2, 3, 4, 5, 6, 7, 8, 0]

print("Solving using standard Depth-First Search (DFS):")
std_dfs_solution = standard_dfs_solve(initial_state, goal_state)

if std_dfs_solution:
    for step, (move, board) in enumerate(std_dfs_solution, 1):
        print(f"Step {step}: Move 0 {move}")
        print_board(board)
else:
    print("No solution found using standard DFS within the depth limit.")
```
