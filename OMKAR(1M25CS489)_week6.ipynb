```python

from itertools import product

def implies(a, b):
    return not a or b

def check_entailment(kb, query, models):
    for model in models:
        if kb(model) and not query(model):
            return False
    return True

symbols = ['P', 'Q', 'R']
models = list(product([True, False], repeat=3))

print("P\tQ\tR\tQ->P\tP->~Q\tQ|R\tKB")

def kb(m):
    P, Q, R = m
    return implies(Q, P) and implies(P, not Q) and (Q or R)

for m in models:
    P, Q, R = m
    a = implies(Q, P)
    b = implies(P, not Q)
    c = Q or R
    result = kb(m)

    print(f"{P}\t{Q}\t{R}\t{a}\t{b}\t{c}\t{result}")

print("\nModels where KB is TRUE:")
for m in models:
    if kb(m):
        print(m)

queries = {
    "R": lambda m: m[2],
    "R -> P": lambda m: implies(m[2], m[0]),
    "Q -> R": lambda m: implies(m[1], m[2])
}

print("\nEntailment Results:")
for name, query in queries.items():
    if check_entailment(kb, query, models):
        print("KB entails", name, ": TRUE")
    else:
        print("KB entails", name, ": FALSE")
```
