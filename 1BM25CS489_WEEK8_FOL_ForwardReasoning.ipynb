```python
print("USN: 1BM25CS489")
print("NAME: OMKAR SADASHIV KOKATNOOR")
print("")
print("FORWARD CHAINING IN FIRST ORDER LOGIC")
print()

# Initial facts
facts = [
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
]

# Forward chaining
changed = True

while changed:
    changed = False

    for fact in facts[:]:

        # Rule 1: All missiles are weapons
        if fact[0] == "Missile":
            new_fact = ("Weapon", fact[1])
            if new_fact not in facts:
                facts.append(new_fact)
                changed = True

        # Rule 2: Enemies are hostile
        if fact[0] == "Enemy":
            new_fact = ("Hostile", fact[1])
            if new_fact not in facts:
                facts.append(new_fact)
                changed = True

    # Rule 3: Robert sells missiles owned by A
    for fact in facts[:]:
        if fact[0] == "Missile":
            missile = fact[1]
            if ("Owns", "A", missile) in facts:
                new_fact = ("Sells", "Robert", missile, "A")
                if new_fact not in facts:
                    facts.append(new_fact)
                    changed = True

    # Rule 4: An American selling weapons to hostile
    # nations is a criminal
    if (("American", "Robert") in facts and
        ("Weapon", "T1") in facts and
        ("Sells", "Robert", "T1", "A") in facts and
        ("Hostile", "A") in facts):

        new_fact = ("Criminal", "Robert")
        if new_fact not in facts:
            facts.append(new_fact)
            changed = True

# Display all facts
print("Knowledge Base:")
for fact in facts:
    print(fact)

# Check goal
if ("Criminal", "Robert") in facts:
    print("\nGoal Proved: Robert is a Criminal")
else:
    print("\nGoal Not Proved")
```
